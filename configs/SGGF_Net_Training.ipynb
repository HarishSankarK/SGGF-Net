```python
# ============================================================================
# STEP 2-5: Setup, Verify Dataset, and Start Training
# ============================================================================

# Step 1: Verify we're in the right directory
import os
print("=" * 70)
print("STEP 1: Verifying directory...")
print("=" * 70)
print(f"Current directory: {os.getcwd()}")
print(f"Contents: {os.listdir('.')}")

# Step 2: Setup and Install Dependencies
print("\n" + "=" * 70)
print("STEP 2: Setting up Environment...")
print("=" * 70)

# Detect environment (same as Cell 0)
IN_COLAB = False
IN_KAGGLE = False
try:
    import google.colab
    IN_COLAB = True
except ImportError:
    if 'google.colab' in sys.modules:
        IN_COLAB = True

if not IN_COLAB:
    if os.environ.get('KAGGLE_KERNEL_TYPE') or os.environ.get('KAGGLE_KERNEL_RUN_TYPE'):
        IN_KAGGLE = True
    elif os.path.exists('/kaggle/working') or os.path.exists('/kaggle/input'):
        IN_KAGGLE = True

ENVIRONMENT = 'Colab' if IN_COLAB else ('Kaggle' if IN_KAGGLE else 'Local')

# Install dependencies only if needed (Colab, Kaggle, or if packages missing)
if IN_COLAB or IN_KAGGLE:
    if IN_COLAB:
        print("📋 Colab: Make sure GPU runtime is enabled:")
        print("   1. Go to: Runtime → Change runtime type")
        print("   2. Select: GPU (T4) or CPU")
        print("   3. Click: Save")
        print("   4. Runtime → Restart runtime")
        print("")
        print("Installing PyTorch and dependencies in Colab...")
        %pip install torch torchvision --index-url https://download.pytorch.org/whl/cu118
    else:  # Kaggle
        print("📋 Kaggle: GPU is automatically enabled if selected in notebook settings")
        print("   To enable GPU: Notebook settings → Accelerator → GPU")
        print("")
        print("Installing PyTorch and dependencies in Kaggle...")
        # Kaggle usually has PyTorch pre-installed, but we'll ensure it's the right version
        %pip install torch torchvision --index-url https://download.pytorch.org/whl/cu118 --quiet
    
    print("\nInstalling other dependencies...")
    %pip install numpy pillow opencv-python tqdm matplotlib scipy --quiet
else:
    print("Local system: Checking if dependencies are installed...")
    required_packages = ['torch', 'torchvision', 'numpy', 'PIL', 'cv2', 'tqdm', 'matplotlib', 'scipy']
    missing_packages = []
    
    for pkg in required_packages:
        try:
            if pkg == 'PIL':
                __import__('PIL')
            elif pkg == 'cv2':
                __import__('cv2')
            else:
                __import__(pkg)
        except ImportError:
            missing_packages.append(pkg)
    
    if missing_packages:
        print(f"⚠ Missing packages: {missing_packages}")
        print("")
        print("Installing missing packages automatically...")
        print("(This may take a few minutes)")
        
        # Install PyTorch first (with CUDA support if available)
        try:
            import subprocess
            # Try to detect if CUDA is available on system
            cuda_available = False
            try:
                result = subprocess.run(['nvcc', '--version'], capture_output=True, text=True)
                if result.returncode == 0:
                    cuda_available = True
                    print("✓ CUDA detected on system - installing CUDA-enabled PyTorch")
            except FileNotFoundError:
                pass
            
            if cuda_available:
                # Try CUDA 11.8 first (common)
                print("Installing PyTorch with CUDA 11.8 support...")
                %pip install torch torchvision --index-url https://download.pytorch.org/whl/cu118
            else:
                # CPU-only PyTorch
                print("Installing CPU-only PyTorch...")
                print("(For GPU support, install CUDA toolkit and reinstall PyTorch)")
                %pip install torch torchvision
        except Exception as e:
            print(f"⚠ PyTorch installation failed: {e}")
            print("  Please install manually: pip install torch torchvision")
            raise
        
        # Install other dependencies
        print("\nInstalling other dependencies...")
        %pip install numpy pillow opencv-python tqdm matplotlib scipy
        
        print("\n✓ Package installation completed!")
        print("  Note: You may need to restart kernel and re-run this cell if you get import errors")
    else:
        print("✓ All required packages are installed")

# Verify device setup (works for both Colab and local)
print("\nChecking device availability...")
# Import torch after installation is complete
try:
    import torch
except ImportError:
    print("⚠ torch not found after installation")
    print("  Please restart kernel and re-run this cell, or install manually:")
    print("  pip install torch torchvision")
    raise

USE_GPU = torch.cuda.is_available()
if USE_GPU:
    gpu_name = torch.cuda.get_device_name(0)
    gpu_memory = torch.cuda.get_device_properties(0).total_memory / (1024**3)  # GB
    print(f'✓ GPU detected: {gpu_name}')
    print(f'✓ GPU Memory: {gpu_memory:.1f} GB')
    if IN_COLAB and 'T4' in gpu_name:
        print('✓ T4 GPU detected - optimal for this training!')
    elif IN_KAGGLE:
        print(f'✓ Using Kaggle GPU: {gpu_name}')
    elif not IN_COLAB and not IN_KAGGLE:
        print('✓ Using local GPU')
else:
    print('⚠ GPU not available - will use CPU')
    if IN_COLAB:
        print('  💡 In Colab: Enable GPU runtime for faster training')
    elif IN_KAGGLE:
        print('  💡 In Kaggle: Enable GPU in notebook settings (Settings → Accelerator → GPU)')
    else:
        print('  💡 Local: Install CUDA-enabled PyTorch for GPU support')

print(f'\n✓ Device setup complete! Will use: {"GPU (CUDA)" if USE_GPU else "CPU"}')

# Step 3: Verify Dataset
print("\n" + "=" * 70)
print("STEP 3: Verifying dataset...")
print("=" * 70)

if os.path.exists('data/hit-uav'):
    print("✓ Dataset found in data/hit-uav/")
    try:
        train_img_dir = 'data/hit-uav/images/train'
        train_label_dir = 'data/hit-uav/labels/train'
        if os.path.exists(train_img_dir):
            num_images = len([f for f in os.listdir(train_img_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))])
            print(f"  Images: {num_images} training images")
        else:
            print(f"  ⚠ Images directory not found: {train_img_dir}")
            
        if os.path.exists(train_label_dir):
            num_labels = len([f for f in os.listdir(train_label_dir) if f.endswith('.txt')])
            print(f"  Labels: {num_labels} training labels")
        else:
            print(f"  ⚠ Labels directory not found: {train_label_dir}")
            
        print("\nDataset structure:")
        !ls -la data/hit-uav/
    except Exception as e:
        print(f"⚠ Error checking dataset: {e}")
        print("  But dataset directory exists, continuing...")
else:
    print("⚠ Dataset not found. Make sure you've pushed it to GitHub.")

# Step 4: Setup checkpoint directory and start training
print("\n" + "=" * 70)
print("STEP 4: Starting Training...")
print("=" * 70)

# Setup checkpoint directory based on environment
if IN_COLAB:
    checkpoint_dir = '/content/drive/MyDrive/SGGF-Net-checkpoints'
    print("✓ Running in Colab - checkpoints will be saved to Google Drive")
elif IN_KAGGLE:
    checkpoint_dir = '/kaggle/working/checkpoints'
    print("✓ Running in Kaggle - checkpoints will be saved to /kaggle/working/")
    print("  Note: Files in /kaggle/working/ persist and can be downloaded")
else:
    checkpoint_dir = os.path.join(os.getcwd(), 'checkpoints')
    print(f"✓ Running locally - checkpoints will be saved to: {checkpoint_dir}")

os.makedirs(checkpoint_dir, exist_ok=True)
print(f'✓ Checkpoint directory: {checkpoint_dir}')

# Final device check before training
print('')
print('🔍 Final device check before training...')
import torch

USE_GPU = torch.cuda.is_available()
if USE_GPU:
    gpu_name = torch.cuda.get_device_name(0)
    print(f'✓ Device ready: {gpu_name} (GPU)')
    print('✓ Training will use GPU (fast!)')
    device_arg = 'cuda'
else:
    print('⚠ GPU not available - will use CPU')
    if IN_COLAB:
        print('  💡 For faster training, enable GPU runtime in Colab')
    else:
        print('  💡 For faster training, install CUDA-enabled PyTorch')
    device_arg = 'cpu'

# ADAPTIVE SETTINGS BASED ON DEVICE
# Settings automatically adjust based on GPU/CPU availability

if USE_GPU:
    # GPU Settings (optimized for T4 or similar)
    batch_size = 1
    max_size = 800
    grad_accum_steps = 4
    use_amp = True
    print("\n" + "=" * 70)
    print("TRAINING CONFIGURATION (GPU Optimized):")
    print("=" * 70)
    print("Device: GPU")
    print(f"Batch size: {batch_size}")
    print(f"Gradient accumulation: {grad_accum_steps} (effective batch_size = {batch_size * grad_accum_steps})")
    print(f"Max image size: {max_size}")
    print("Mixed Precision: Enabled (AMP)")
else:
    # CPU Settings (conservative to avoid OOM)
    batch_size = 1
    max_size = 640  # Smaller for CPU
    grad_accum_steps = 2  # Less accumulation for CPU
    use_amp = False  # AMP not supported on CPU
    print("\n" + "=" * 70)
    print("TRAINING CONFIGURATION (CPU - Slower but Stable):")
    print("=" * 70)
    print("Device: CPU")
    print(f"Batch size: {batch_size}")
    print(f"Gradient accumulation: {grad_accum_steps} (effective batch_size = {batch_size * grad_accum_steps})")
    print(f"Max image size: {max_size}")
    print("Mixed Precision: Disabled (not supported on CPU)")
    print("⚠ Training on CPU will be slower - consider using GPU if available")

print("Learning rate: 0.0005")
print(f"Checkpoint directory: {checkpoint_dir}")
print("=" * 70)
print("")

# Detect Python command (python3 on macOS/Linux, python on Windows/Colab)
import shutil
if shutil.which('python3'):
    python_cmd = 'python3'
elif shutil.which('python'):
    python_cmd = 'python'
else:
    # Fallback: use sys.executable (the current Python interpreter)
    import sys
    python_cmd = sys.executable
    print(f"⚠ Using current interpreter: {python_cmd}")

print(f"✓ Python command: {python_cmd}")

# Build training command
amp_flag = "--use_amp" if use_amp else "--no_amp"
train_cmd = f"""{python_cmd} scripts/train.py \\
    --dataset hituav \\
    --data_dir data/hit-uav \\
    --num_classes 6 \\
    --batch_size {batch_size} \\
    --grad_accum_steps {grad_accum_steps} \\
    --num_epochs 50 \\
    --lr 0.0005 \\
    --max_size {max_size} \\
    --warmup_epochs 5 \\
    --checkpoint_dir {checkpoint_dir} \\
    --device {device_arg} \\
    {amp_flag} \\
    --val_freq 10"""

print("\nStarting training with command:")
print(train_cmd)
print("")

# Start training using subprocess (allows variable substitution)
import subprocess
import sys

train_args = [
    python_cmd,
    'scripts/train.py',
    '--dataset', 'hituav',
    '--data_dir', 'data/hit-uav',
    '--num_classes', '6',
    '--batch_size', str(batch_size),
    '--grad_accum_steps', str(grad_accum_steps),
    '--num_epochs', '50',
    '--lr', '0.0005',
    '--max_size', str(max_size),
    '--warmup_epochs', '5',
    '--checkpoint_dir', checkpoint_dir,
    '--device', device_arg,
    amp_flag,
    '--val_freq', '10'
]

print(f"Executing: {' '.join(train_args)}")
print("")

# Execute training
try:
    subprocess.run(train_args, check=True)
except subprocess.CalledProcessError as e:
    print(f"Training failed with exit code {e.returncode}")
    sys.exit(e.returncode)
except FileNotFoundError:
    print(f"Error: {python_cmd} not found!")
    print("Please ensure Python is installed and in PATH")
    print(f"Tried: {python_cmd}")
    raise

print("\n" + "=" * 70)
print("Training completed!")
print("=" * 70)
```

```python
# Resume training from a checkpoint (works on both Colab and local)
import os
import sys

# Detect environment
IN_COLAB = False
IN_KAGGLE = False
try:
    import google.colab
    IN_COLAB = True
except ImportError:
    if 'google.colab' in sys.modules:
        IN_COLAB = True

if not IN_COLAB:
    if os.environ.get('KAGGLE_KERNEL_TYPE') or os.environ.get('KAGGLE_KERNEL_RUN_TYPE'):
        IN_KAGGLE = True
    elif os.path.exists('/kaggle/working') or os.path.exists('/kaggle/input'):
        IN_KAGGLE = True

ENVIRONMENT = 'Colab' if IN_COLAB else ('Kaggle' if IN_KAGGLE else 'Local')

# Setup checkpoint directory based on environment
if IN_COLAB:
    checkpoint_dir = '/content/drive/MyDrive/SGGF-Net-checkpoints'
elif IN_KAGGLE:
    checkpoint_dir = '/kaggle/working/checkpoints'
else:
    checkpoint_dir = os.path.join(os.getcwd(), 'checkpoints')

print(f'✓ Environment: {ENVIRONMENT}')
print(f'✓ Checkpoint directory: {checkpoint_dir}')

# Check if checkpoint exists
latest_checkpoint = os.path.join(checkpoint_dir, 'latest.pth')
best_checkpoint = os.path.join(checkpoint_dir, 'best.pth')

if os.path.exists(latest_checkpoint):
    resume_from = latest_checkpoint
    print(f'✓ Resuming from: {resume_from}')
elif os.path.exists(best_checkpoint):
    resume_from = best_checkpoint
    print(f'✓ Resuming from: {resume_from}')
else:
    resume_from = None
    print('⚠ No checkpoint found, starting fresh training')

# Resume training (adaptive settings based on device)
import torch
USE_GPU = torch.cuda.is_available()
device_arg = 'cuda' if USE_GPU else 'cpu'

# Detect Python command
import shutil
if shutil.which('python3'):
    python_cmd = 'python3'
elif shutil.which('python'):
    python_cmd = 'python'
else:
    python_cmd = sys.executable

if USE_GPU:
    batch_size = 1
    max_size = 800
    grad_accum_steps = 4
    amp_flag = '--use_amp'
else:
    batch_size = 1
    max_size = 640
    grad_accum_steps = 2
    amp_flag = '--no_amp'

if resume_from:
    print(f'✓ Device: {device_arg.upper()}')
    print(f'✓ Batch size: {batch_size}, Max size: {max_size}')
    print(f'✓ Python command: {python_cmd}')
    print('')
    
    # Resume training using subprocess
    import subprocess
    import sys
    
    resume_args = [
        python_cmd,
        'scripts/train.py',
        '--dataset', 'hituav',
        '--data_dir', 'data/hit-uav',
        '--num_classes', '6',
        '--batch_size', str(batch_size),
        '--grad_accum_steps', str(grad_accum_steps),
        '--num_epochs', '50',
        '--lr', '0.0005',
        '--max_size', str(max_size),
        '--warmup_epochs', '5',
        '--checkpoint_dir', checkpoint_dir,
        '--resume', resume_from,
        '--device', device_arg,
        '--val_freq', '10',
        amp_flag
    ]
    
    print(f"Executing: {' '.join(resume_args)}")
    print("")
    
    try:
        subprocess.run(resume_args, check=True)
    except subprocess.CalledProcessError as e:
        print(f"Resume training failed with exit code {e.returncode}")
        sys.exit(e.returncode)
    except FileNotFoundError:
        print(f"Error: {python_cmd} not found!")
        print(f"Tried: {python_cmd}")
        raise
else:
    print('⚠ No checkpoint to resume from. Run Cell 2 (Step 4) to start training.')
```
